```python
def idGroup():
    ids = {0: [], 5:[], 10:[]}

    for minute in [0, 5, 10]:
        path = f'./Data/{minute}m break/'
        filelist = os.listdir(path)
        ids[minute] = list(set([int(file[:2]) for file in filelist]))
        
    return ids
```

```python
def DataParser(minute):
    path = f'./Data/{minute}m break/'
    filelist = os.listdir(path)
    ids = set([file[:2] for file in filelist])
    group = {}

    for idx in ids:
        data = []
        for i in range(3):
            data.append(pd.read_csv(f'./Data/{minute}m break/{idx}_{i+1}.csv', index_col=0).T.reset_index())
            data[i]['index'] = data[i]['index'].astype(int)
            data[i]['class'] = data[i]['class'].astype(int)
            data[i]['speed'] = data[i]['speed'].astype(float)
            data[i]['text'] = data[i]['text']
            data[i]['color'] = data[i]['color'].apply(lambda x: list(map(int, x.strip('()').split(', '))))
        data.append(data[0])
        data[3] = pd.concat([data[3], data[1], data[2]], axis=0, ignore_index=True)
        data[3] = data[3].iloc[:, 1:].reset_index()
        group[idx] = data
    
    return group
```

```python
Group = {}
for i in [0, 5, 10]:
    Group[i] = DataParser(i)
```

# 1
### 전체 데이터에 대한 반응속도 추세선의 기울기 간의 상대 점수

#### a. 전체에 대해

```python
def all_gradient_z_score():
    gradient = {}
    for g in Group.values():
        for k, p in g.items():
            speed = p[3]['speed']
    
            x = np.arange(len(speed))
            y = speed
    
            coeffs = np.polyfit(x, y, 1)

            slope = coeffs[0]
            gradient[k] = slope
    gradient_np = np.array(list(gradient.values()))

    mean_slope = np.mean(gradient_np)
    std_slope = np.std(gradient_np)
    z_score = (gradient_np - mean_slope) / std_slope
    
    zperid = {}
    for idx, score in zip(gradient.keys(), z_score):
        zperid[idx] = round(score, 3)

    sorted_zscore = {k:v for k, v in sorted(zperid.items())}

    df = pd.DataFrame(list(sorted_zscore.items()), columns = ['id', 'z_score'])
    df['rank'] = df['z_score'].rank(ascending=False, method='min').astype(int)
    df.set_index('id', inplace=True)
    df.to_excel('./Graphs/z_score_all.xlsx')
    return df
```

```python
all_gradient_z_score()
```

#### b. 그룹 별

```python
def Group_gradient_z_score():
    for m, g in Group.items():
        gradient = {}
        for k, p in g.items():
            speed = p[3]['speed']
    
            x = np.arange(len(speed))
            y = speed
    
            coeffs = np.polyfit(x, y, 1)

            slope = coeffs[0]
            gradient[k] = slope
        gradient_np = np.array(list(gradient.values()))
    
        mean_slope = np.mean(gradient_np)
        std_slope = np.std(gradient_np)
        z_score = (gradient_np - mean_slope) / std_slope
        
        zperid = {}
        for idx, score in zip(gradient.keys(), z_score):
            zperid[idx] = round(score, 3)
    
        sorted_zscore = {k:v for k, v in sorted(zperid.items())}
    
        df = pd.DataFrame(list(sorted_zscore.items()), columns = ['id', 'z_score'])
        df['rank'] = df['z_score'].rank(ascending=False, method='min').astype(int)
        df.set_index('id', inplace=True)
        df.to_excel(f'./Graphs/z_score_group_{m}.xlsx')
```

### scatter plot

```python
def draw_plot():
    for g in Group.values():
        for k, p in g.items():
            speed = p[3]['speed']
    
            x = np.arange(len(speed))
            y = speed
    
            coeffs = np.polyfit(x, y, 1)
            poly = np.poly1d(coeffs)

            slope = coeffs[0]

            plt.scatter(x, y, alpha=0.7, s = 2)

            plt.plot(x, poly(x), linestyle='-', c = 'orange')

            plt.xlabel("Time")
            plt.ylabel("Speed")
            plt.title(f"{k} Speed Graph")

            plt.text(0.7 * max(x), 0.9 * max(y), f"Slope: {slope:.6f}", fontsize=10, color='red')
            
            plt.savefig(f"./Graphs/All/{k}.png")

            plt.clf()
```

```python
def colors(color):
    rgb = []
    if color == '빨강':
        rgb = [255, 0, 0]
    elif color == '초록':
        rgb = [0, 255, 0]
    elif color == '파랑':
        rgb = [0, 0, 255]
    elif color == '노랑':
        rgb = [255, 255, 0]
    return rgb
```

#### 1. congurent 여부에 따른 반응속도 분포와 평균
Neutral/Congruent/Incongruent

```python
speed_by_color = {'congruent': [], 'incongruent': [], 'neutral': []}
for m, g in Group.items():
    for k, p in g.items():
        df = pd.DataFrame(p[3])
        speed_by_color['neutral'].extend(df.loc[df['text'] == 'XXXX', 'speed'].to_list())
        speed_by_color['congruent'].extend(df.loc[df['text'].apply(colors) == df['color'], 'speed'].to_list())
        speed_by_color['incongruent'].extend(df.loc[(df['text'].apply(colors) != df['color']) & (df['text'] != 'XXXX'), 'speed'].to_list())
```

```python
max_length = max(len(values) for values in speed_by_color.values())

for key, values in speed_by_color.items():
    while len(values) < max_length:
        values.append(np.nan)

df = pd.DataFrame(speed_by_color)

df_melted = df.melt(var_name='Condition', value_name='Speed')

plt.figure(figsize=(8, 6))

ax = sns.boxplot(x='Condition', y='Speed', data=df_melted, 
                 hue='Condition',         # 'Condition'을 hue로 설정
                 palette='Set2',          # 색상 팔레트 지정
                 width=0.4)               # 수염 그래프 너비 조정

for condition in df_melted['Condition'].unique():
    mean_value = df_melted[df_melted['Condition'] == condition]['Speed'].mean()

    ax.text(x=df_melted['Condition'].unique().tolist().index(condition), 
            y=mean_value + 0.13,  
            s=f'Mean: {mean_value:.2f}',  
            ha='center',  
            fontsize=12,   
            color='black') 

plt.title('Reaction time by Condition')
plt.xlabel('Condition')
plt.ylabel('Time taken')

plt.savefig(f"./Graphs/Boxplot_Speed_by_Condition.png")

plt.show()
```

2. “당신은 몇 분 이상 휴식할 때 승률이 좋습니다.”

#### 3) timeline data -> 분당 골드 획득 변동
1. 선그래프
2. 4개 연속을 그냥 좌우 concat
3. concat한 지점 표시(수직선)
4. 4게임이 안 나오는 사람은 수를 줄여서

> 실험대상: 일 희 #1 2

```python
load_dotenv()
KEY = os.environ.get('RIOT_API_KEY')
```

```python
def collect_timelinedata(ID, Tagline):
    puuid = API.gameNameTopuuid(ID, Tagline, KEY)['puuid']
    matchids = API.puuidTomatchId(puuid, KEY)
    
    pbar = tqdm(matchids,
               total = len(matchids),
               desc = 'Turnning Match ID to Match Data',
               ncols = 100,
               leave = True)

    gold_per_time_data = []
    for matchid in pbar:
        data = API.timeline(matchid, KEY)['info']
        playerid = 0
        for participant in data['participants']:
            if participant['puuid'] == puuid:
                playerid = participant['participantId']
        if playerid == 0:
            print('Can\'t find puuid')
            continue

        frames = data['frames']
        gameinfodict = {'start':0, 'end':0, 'gold_flow':[]}
        
        try:
            gameinfodict['start'] = [event['realTimestamp'] for event in frames[0]['events'] if event['type'] == 'PAUSE_END'][0]
            gameinfodict['end'] = [event['realTimestamp'] for event in frames[-1]['events'] if event['type'] == 'GAME_END'][0]
        except Exception as e:
            print(e)
            continue

        try:
            previous_gold = 500
            for frame in frames:
                current_gold = frame['participantFrames'][f'{playerid}']['totalGold']
                gameinfodict['gold_flow'].append(current_gold-previous_gold)
                previous_gold = current_gold
        except Exception as e:
            print(e)
            continue
        gold_per_time_data.append(gameinfodict)
    try:
        with open(f'./Data/Subject gold flow using API timeline data/{ID}#{Tagline}.json', 'w') as f:
            json.dump(gold_per_time_data, f, indent=4)
        print('Saved as json file')
        
    except Exception as e:
        print(e)
```

```python
collect_timelinedata('일 희', '1 2')
```
